```python
import os
import pandas as pd
import re

def summarize_card_values(folder_path, prefix_pattern=r'^(weekly|monthly)_data_aggregated_'):
    """
    Compute mean and median prices for each card file in the folder.
    Cleans file names to keep only the card name.
    Returns two DataFrames: sorted by mean, and sorted by median.
    """
    results = []

    for file in os.listdir(folder_path):
        if file.endswith(".xlsx") or file.endswith(".csv"):  
            file_path = os.path.join(folder_path, file)
            
            # Load file
            if file.endswith(".xlsx"):
                df = pd.read_excel(file_path)
            else:
                df = pd.read_csv(file_path)
            
            # Clean numeric values
            prices = pd.to_numeric(df['avg_price'], errors='coerce').dropna()
            
            if not prices.empty:
                mean_price = prices.mean()
                median_price = prices.median()
                
                # Clean the file name to show just the card
                card_name = re.sub(prefix_pattern, '', file)  # remove weekly/monthly prefix
                card_name = card_name.rsplit('.', 1)[0]       # drop extension
                
                results.append({
                    "card": card_name,
                    "mean_price": mean_price,
                    "median_price": median_price
                })
    
    results_df = pd.DataFrame(results)
    
    sorted_by_mean = results_df.sort_values(by="mean_price", ascending=False).reset_index(drop=True)
    sorted_by_median = results_df.sort_values(by="median_price", ascending=False).reset_index(drop=True)
    
    return sorted_by_mean, sorted_by_median



# Example usage:
weekly_folder = r"TIME_SERIES_ANALYSIS\EFFICIENCY_INDEX_PRICES_DATA_raw\weekly"
monthly_folder = r"TIME_SERIES_ANALYSIS\EFFICIENCY_INDEX_PRICES_DATA_raw\monthly"

weekly_mean, weekly_median = summarize_card_values(weekly_folder)
monthly_mean, monthly_median = summarize_card_values(monthly_folder)

print("=== Weekly sorted by mean ===")
print(weekly_mean)
#print("\n=== Weekly sorted by median ===")
#print(weekly_median)

print("\n=== Monthly sorted by mean ===")
print(monthly_mean)
#print("\n=== Monthly sorted by median ===")
#print(monthly_median)
```
